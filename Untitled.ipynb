```python
primo = 1 
secondo = 1 
final = primo + secondo 
# how many fibo values to be added  => n 
n = 4

for i in range(n-2):
    fibo = primo + secondo 
    primo = secondo
    secondo = fibo
    final += fibo 


print(final)


```

```python
def calculate_bmi(weight, height):
    height /= 100
    bmi = weight/((height)**2)
    return bmi
```

```python
calculate_bmi(58, 170)
```

```python
for member in ["haniye", "milad", "mojgan"]:
    member_weight = int(input("please enter the weight of {}".format(member)))
    member_height = int(input("please enter the height of {}".format(member)))
    bmi = calculate_bmi(member_weight, member_height)
    print("bmi for {} is {}".format(member, bmi))
```
